```python
from pathlib import Path
data_yaml = "data/cowboy/yolo/data.yaml"
```

```python
from ultralytics import YOLO

model = YOLO("yolo26n.pt")  # 小型预训练模型，适合先跑通流程
```

```python
results = model.train(
    data=str(data_yaml),
    epochs=100,
    imgsz=640
)
```

```python
best_model = YOLO(r"runs\detect\train\weights\best.pt")
```

```python
from PIL import Image
from IPython.display import display

image_path = r"data/cowboy/images/0a68787ee0f20c6d.jpg"

results = best_model.predict(
    source=image_path,
    conf=0.25,
    save=True
)

display(Image.fromarray(results[0].plot()[:, :, ::-1]))
```
