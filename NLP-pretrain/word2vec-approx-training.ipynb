### 词嵌入
one-hot vec会丢失信息之间的关联以及词数量较大时数组维数过大
因此2013年mikolov提出了两种自监督的word to vector的模型
1. 跳元模型（skip-gram）

2. 连续词袋（CBOW, continues bags of words）


跳元模型主要使用中心词来生成临近词


连续词袋主要使用临近词来生成中心词


### 负采样

不再完整计算对整张词表的打分，而是只计算正例和挑选的K个负例。
训练时让正例的点积变大，让负例的点积变小。每个正例只需再处理 \(K\) 个随机负例，所以每步的计算量大致取决于 \(K\)，而不用扫完整个词表。

### 层序Softmax

树提前建好；训练时借树快速预测词；训练完成后主要拿词向量用。
